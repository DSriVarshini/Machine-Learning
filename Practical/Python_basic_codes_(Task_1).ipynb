```python
num = 456
rev = 0

while num > 0:
  digit = num % 10
  rev = rev*10+digit
  num = num//10
print("Reversed Number: ",rev)




```

```python
num = 6

if num <= 1:
    print(num, "is not a prime number")
else:
    is_prime = True
    i = 2

    while i < num:
        if num % i == 0:
            is_prime = False
            break
        i += 1

    if is_prime:
        print(num, "is a prime number")
    else:
        print(num, "is not a prime number")
```
